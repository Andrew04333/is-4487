## Getting data into Pandas

In this case we will load data from the statsmodels.org library.  The data is built in to a Python library, so we don't need to export it from an external source.  In future labs, we will pull the data from GitHub or other sources.  

```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```

### 🔧 Do the following: Task 1 - Work with DataFrame

🔧 Assignment Tasks - Practice selecting specific rows and columns.
- 1.1 Extract the mpg (miles per gallon) column and calculate its mean.
- 1.2 Create a new dataframe containing only the rows where mpg is greater than 25.
- 1.3 Create a new dataframe with rows with the number of cylinders (cyl) is 4.  Include only the columns mpg and hp.



```python
# 1.1 Extract the mpg (miles per gallon) column, then calculate its mean.
mpg = df['mpg']                    # square brackets + a column name pulls out one column
display(mpg)                       # the extracted column on its own
print('Mean mpg:', mpg.mean())     # the average across all 32 cars
```

```python
# 1.2 New dataframe containing only the rows where mpg is greater than 25.
df_high_mpg = df[df['mpg'] > 25]   # the inner test is True/False for every row; only True rows are kept
print('Rows kept:', len(df_high_mpg), 'out of', len(df))
df_high_mpg
```

```python
# 1.3 New dataframe: rows where cyl is 4, keeping only the mpg and hp columns.
df_4cyl = df.loc[df['cyl'] == 4, ['mpg', 'hp']]   # .loc[which rows, which columns]
print('Shape (rows, columns):', df_4cyl.shape)
df_4cyl
```


### 🔧 Do the following - Task 2 - Introduce derived columns.
- 2.1 Create a new column power_to_weight_ratio as hp / wt (horsepower divided by weight).

```python
# 2.1 Derived column: horsepower per 1,000 lbs of weight.
df['power_to_weight_ratio'] = df['hp'] / df['wt']
df[['hp', 'wt', 'power_to_weight_ratio']].head(10)
```

### 🔧 Do the following - Task 3 - Summarize data.
- 3.1 Group the data by the number of cylinders (cyl) and calculate the average mpg for each group.
- 3.2 Find the maximum horsepower (hp) for cars in each gear group.

```python
# 3.1 Group by number of cylinders, then average mpg within each group.
avg_mpg_by_cyl = df.groupby('cyl')['mpg'].mean()
avg_mpg_by_cyl
```

```python
# 3.2 Maximum horsepower within each gear group.
max_hp_by_gear = df.groupby('gear')['hp'].max()
max_hp_by_gear
```

### 🔧 Do the following - Task 4 - Sorting Data
- 4.1 Sort the dataframe by mpg in descending order.
- 4.2 Sort the dataframe first by cyl and then by hp.

```python
# 4.1 Sort the dataframe by mpg, highest first.
df_by_mpg = df.sort_values('mpg', ascending=False)
df_by_mpg
```

```python
# 4.2 Sort first by cyl, then by hp within each cylinder group.
df_by_cyl_hp = df.sort_values(['cyl', 'hp'])
df_by_cyl_hp
```
